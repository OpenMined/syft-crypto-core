```python
import syft_crypto_python as syc
```

```python
# Alice material
mat = syc.generate_identity_material("alice@example.com")
alice_keys = syc.SyftRecoveryKey.from_hex_string(mat.recovery_key_hex).derive_keys()
alice_pub = alice_keys.to_public_bundle()
```

```python
alice_pub
```

```python
# Bob
bob_keys = syc.SyftRecoveryKey.generate().derive_keys()
bob_pub = bob_keys.to_public_bundle()
```

```python
env_bytes = syc.encrypt_message(
  "alice@example.com",
  alice_keys,
  [syc.EncryptionRecipient("bob@example.com", bob_pub)],
  b"hello bob!",
  filename_hint="msg.txt",
)
```

```python
parsed = syc.parse_envelope(env_bytes)
syc.verify_envelope_signature(parsed, alice_pub.identity_key_bytes)
```

```python
pt = syc.decrypt_message("bob@example.com", bob_keys, alice_pub, parsed)
assert pt == b"hello bob!"
print("encrypt/decrypt OK")
```
